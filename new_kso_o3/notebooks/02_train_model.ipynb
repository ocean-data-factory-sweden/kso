# 02 – Train model
Loads YAML config, instantiates trainer, and starts training loop. Edit parameters below to switch backend or dataset paths.

```python
# ⚙️ User parameters
MODEL_BACKEND = 'pytorch'  # pytorch | yolo | hf
CONFIG_DIR = 'configs'
DATA_DIR = 'data'  # folder containing train/validation (or yolo_data.yaml for YOLO)
```

```python
from pathlib import Path
import importlib
from kso.config import load_project_config
from kso.registry import get_trainer

# Load config and override backend if needed
cfg = load_project_config(CONFIG_DIR)
if MODEL_BACKEND and cfg.model != MODEL_BACKEND:
    cfg_dict = cfg.model_dump()
    cfg_dict['model'] = MODEL_BACKEND
    cfg = cfg.__class__(**cfg_dict)

# Ensure trainer registered
importlib.import_module(f'kso.trainers.{cfg.model}')
trainer_cls = get_trainer(cfg.model)
trainer = trainer_cls(cfg.model_params)

print('Training with', trainer_cls.__name__)
trainer.train(Path(DATA_DIR))

print('Saving…')
trainer.save(Path(cfg.output_dir))
```
