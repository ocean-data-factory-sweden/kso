# 03 – Evaluate model

```python
# ⚙️ Parameters
MODEL_BACKEND = 'pytorch'  # pytorch | yolo | hf
DATA_DIR = 'data'
```

```python
import importlib
from pathlib import Path
from kso.registry import get_trainer

importlib.import_module(f'kso.trainers.{MODEL_BACKEND}')
TrainerCls = get_trainer(MODEL_BACKEND)
trainer = TrainerCls({})  # empty config just to call evaluate

metric = trainer.evaluate(Path(DATA_DIR))
print('Evaluation metric:', metric)
```
