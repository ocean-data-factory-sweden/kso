# 01 – Download example dataset
This notebook fetches a small flower-classification dataset and unpacks it into `data/`.
Update the variables in the first code cell if you wish to use another URL or destination.

```python
# ⚙️ User-editable parameters
DATASET_URL = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"
DEST_DIR = "data"  # relative to project root
   
```

```python
import subprocess, os, pathlib, shlex
pathlib.Path(DEST_DIR).mkdir(parents=True, exist_ok=True)
cmd = f'curl -L {shlex.quote(DATASET_URL)} | tar -xz --strip-components=1 -C {shlex.quote(DEST_DIR)}'
print(cmd)
subprocess.run(cmd, shell=True, check=True)
```
