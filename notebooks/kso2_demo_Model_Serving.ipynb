## Model inference 

```python
import os
from kso2 import (
    loading_model,
    model_inference,
    save_predictions,
    load_project,
    )
```

```python
project = load_project("Test project 1")
```

```python
model=loading_model(model_name="test model", version=1)
```

```python
import cv2
image="/Users/ghaith/Desktop/kso/kso/projects/test_project_1/coco8/images/val/000000000036.jpg"
img = cv2.imread(image)
img = cv2.resize(img, (640, 640))
dir="/Users/ghaith/Desktop/kso/kso/projects/test_project_1/coco8/images/val"
```

```python
result=model_inference(model=model,data_path=dir)
```

```python
save_predictions(project=project,predictions=result,save_dir="/Users/ghaith/Desktop/kso/kso/projects/test_project_1/f")
```
