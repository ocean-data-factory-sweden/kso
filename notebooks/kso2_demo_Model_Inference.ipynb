## Model inference 

```python
from kso2 import (
    loading_model,
    model_inference,
    save_predictions,
    load_project,
    memory_estimator,
    get_registered_models,
)
```

```python
project = load_project("Test project 1")
```

```python
listed_models = get_registered_models(project)
listed_models
```

```python
model = loading_model(model_name="test model", version=1)
```

## check the resource and estimate the time of the inference

```python
image_path = "/Users/ghaith/Desktop/kso/kso/projects/test_project_1/coco8/images/val/000000000036.jpg"
```

```python
memory_estimator(model, image_path, batch_size=1)
```

```python
result = model_inference(model=model, data_path=image_path)
```

```python
save_predictions(
    project=project,
    predictions=result,
    save_dir="/Users/ghaith/Desktop/kso/kso/projects/test_project_1",
)
```
