```python
project_name = "Template project"  # available projects can be found in "../kso_utils/db_starter/projects_list.csv"
model = "."  # ???
download_dir = "."
upload_title = ""
upload_description = ""
ACCESS_TOKEN = ""  # Zenodo: It is found in API settings by going to your Zenodo profile and clicking on Applications.
archive_dir = ""  # archive to upload
```

### Initiate project's database

```python
# Find project
project = p_utils.find_project(project_name=project_name)
# Initialise mlp
mlp = MLProjectProcessor(project)
```

# Find model and add to Zenodo

### Download the model

```python
artifact_dir = mlp.get_model(model, download_dir)
```

# ZENODO Upload

### Find archive for upload to Zenodo

```python
archive_path = zenodo_utils.zip_folder(archive_dir)
```

### Add archive to upload

```python
depo_id = zenodo_utils.upload_archive(ACCESS_TOKEN, artifact_dir=archive_path)
```

### Indicate database authors together with  their organisations
