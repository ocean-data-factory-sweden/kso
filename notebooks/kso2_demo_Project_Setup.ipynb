# Import necessary libraries and functions from the kso2 package

```python
# These functions handle project creation, data management, model training, and serving.
import os
from kso2 import (
    create_project,
    add_data,
    add_Biigle_data,
    run_augmentation,
)
```

## Create a new project instance.

```python
# This initializes the project structure, creating necessary folders and configuration files.
project = create_project("Test project 1")
```

```python
project.data_path["ultralytics_data_path"]
```

## Add a dataset to the project.

```python
# Calling add_data without arguments adds the default COCO8 dataset for demonstration.
# This prepares the data configuration required for training.
add_data(project=project)
```

## Import data from BIIGLE.

```python
# If you have BIIGLE data, configure the paths below to convert it to YOLO format.
# This allows you to use your own annotated datasets.

BIIGLE_CSV_PATH = "~/desktop/output/biigle_dataset/27737-baltic.csv"  # Path to the BIIGLE CSV annotation file.
IMAGES_ROOT = "~/desktop/output/biigle_dataset/baltic_images"  # Directory containing all BIIGLE image files.
# DATASET_DIR = (
#     "~/Desktop/output"  # Optional: Output folder where the YOLO dataset will be created.
# )

add_Biigle_data(
    project=project,
    BIIGLE_CSV_PATH=BIIGLE_CSV_PATH,
    IMAGES_ROOT=IMAGES_ROOT,
    # DATASET_DIR=DATASET_DIR,
)
```

## OPTIONAL DATA AUGMENTATION

```python
# Controls whether and how dataset augmentation is applied.

# - data_type: selects which dataset to augment
#   ("Biigle_path" or "ultralytics_data_path")
# - AUGMENT_FACTOR: fraction of the dataset to augment (0–1)
# - RANDOM_SEED: ensures reproducibility
data_type = "Biigle_path"  # or "ultralytics_data_path"
augment_factor = 0.5
random_seed = 42

run_augmentation(
    project=project,
    data_type=data_type,
    augment_factor=augment_factor,
    random_seed=random_seed,
)
```
