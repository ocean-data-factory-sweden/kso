# Import necessary libraries and functions from the kso2 package

```python
# These functions handle project creation, data management, model training, and serving.
import os
from kso2 import (
    create_project,
    add_model,
    add_data,
    add_Biigle_data,
)
```

## Create a new project instance.

```python
# This initializes the project structure, creating necessary folders and configuration files.
project = create_project("Test project 1")
```

## Add a dataset to the project.

```python
# Calling add_data without arguments adds the default COCO8 dataset for demonstration.
# This prepares the data configuration required for training.
add_data(project=project)
```

## Import data from BIIGLE.

```python
# If you have BIIGLE data, configure the paths below to convert it to YOLO format.
# This allows you to use your own annotated datasets.

BIIGLE_CSV_PATH = (
    "~/path/biigle_dataset/27737-baltic.csv"  # Path to the BIIGLE CSV annotation file.
)
IMAGES_ROOT = "~/path/biigle_dataset/baltic_images"  # Directory containing all BIIGLE image files.
# DATASET_DIR = (
#     "~/Desktop/output"  # Optional: Output folder where the YOLO dataset will be created.
# )

add_Biigle_data(
    project=project,
    BIIGLE_CSV_PATH=BIIGLE_CSV_PATH,
    IMAGES_ROOT=IMAGES_ROOT,
    # DATASET_DIR=DATASET_DIR,
)
```

## Configure the model to be used for training.

```python
# 'yolov8n.pt' is a nano version of YOLOv8, which is small and fast, making it good for quick tests.
# You can specify other YOLO versions or provide a path to a local model file.
# See available models here: https://docs.ultralytics.com/models/
add_model(project=project, model="yolov8n.pt", model_name="best model")
```
