# Import necessary libraries and functions from the kso2 package

```python
# These functions handle project creation, data management, model training, and serving.
import os
from kso import (
    ProjectManager,
    run_augmentation,
)
```

## Create a new project instance.

```python
# This initializes the project structure, creating necessary folders and configuration files.
proj = ProjectManager()
project = proj.create_project(
    project_name="Test project 3",
    project_path="kso/projects",
    model_name="best model",
    weights_path="yolo11n.pt",
    # ultralytics_data="kso/src/kso/default_dataset/coco8.yaml",
)
```

## Add a dataset to the project.

```python
# Calling add_data without the argument "data_path" adds the default COCO8 dataset for demonstration.
# This prepares the data configuration required for training.
proj.add_data(
    project=project,
    data_type="yolo_dataset",  # Biigle_dataset yolo_dataset
    data_path="kso/projects/data/test_yolo.yaml",
    # images_root="~/desktop/output/biigle_dataset/baltic_images",
    # dataset_dir="~/Desktop",
)
```

## OPTIONAL DATA AUGMENTATION

```python
# Controls whether and how dataset augmentation is applied.

# - data_type: selects which dataset to augment
#   ("Biigle_path" or "ultralytics_data_path")
# - AUGMENT_FACTOR: fraction of the dataset to augment (0–1)
# - RANDOM_SEED: ensures reproducibility
augment_factor = 0.5
random_seed = 42

run_augmentation(
    data_yaml_path="kso/projects/dataset/data.yaml",
    augment_factor=augment_factor,
    random_seed=random_seed,
)
```
