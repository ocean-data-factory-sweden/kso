# ZENODO Upload

### Set the environment variable for the Zenodo ACCESS TOKEN

```python
from dotenv import load_dotenv
import os
from kso import publish_zenodo

load_dotenv()
```

## Set Zenodo access token

```python
ACCESS_TOKEN = os.getenv("zenodo_token")
pub = publish_zenodo(ACCESS_TOKEN=ACCESS_TOKEN)
```

### Upload the zipped file to Zenodo as Draft

```python
pub.upload_folder_zenodo(
    folder_path="kso/projects/test_project_6/mlruns/1b06ae18bdd4419bb8db874a523a9d64.zip"
)
```

### Update the metadata of the draft

```python
pub.update_metadata(
    title="gth",
    upload_type="dataset",
    description="testing the zenodo api",
    creators_name="gth,cc",
    affiliation="none",
)
```

### Publish the draft

```python
pub.publish_file(id=19821847)
```

### Search Records

```python
# search through published records
# Parameters : query author doi title all_versions  sort
pub.record_search(query="DEEP LEARNING MODELS")
```

### Download Record

```python
# download a record based on ID number,
pub.download_record(id=20295836, dir="~/Desktop/kso/projects/records")
```
