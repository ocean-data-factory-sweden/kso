## Model inference 

```python
from kso import (
    ProjectManager,
    TrainingManager,
    MLflowServerManager,
    ModelProfiler,
)
```

```python
proj = ProjectManager()
project = proj.load_project(
    yaml_path="kso/projects/test_project_3/test_project_3.project.yaml"
)
```

```python
serve = MLflowServerManager()
listed_models = serve.get_registered_models(project)
listed_models
```

```python
image_path = "kso/projects/videos"
```

```python
train = TrainingManager()
model = train.loading_model(project=project, model_name="best model", version=1)
```

## check the resource and estimate the time of the inference

```python
prof = ModelProfiler()
prof.memory_estimator(model, image_path, batch_size=16)
```

```python
result = train.model_inference(model=model, data_path=image_path, vid_stride=100)
```

```python
serve.save_predictions(
    project=project,
    predictions=result,
    save_dir="kso/projects/test_project_6",
    output_format="csv",
)
```
